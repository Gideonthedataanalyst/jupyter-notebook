## Initial Place Holder in Numpy Array

### In particular cases, we may wish to initialize numpy arrays with certain values

### Creating a numpy array with all values being zero

#### np.zeros()

```python
### Creating a numpy array with all values being zero

### np.zeros(rows,columns)

d=np.zeros((4,5))
print(d)
```

```python
d.shape
```

```python
# create a numpy array with all values being 1 using np.ones(rows,columns)

e=np.ones((3,4))
print(e)
```

```python
e.shape
```

```python
#array of particular values using np.full((r,c),value)

f=np.full((4,5),8)
print(f)
```

```python
f.shape
```

```python
# Creating an identity matrix i.e all the main diagonal elements will be one and the rest of the values will be zeros
# Note that identity matrices are square matrices i.e the no. of rows=no.of columns
# matrix with same number of columns and rows,in which all the diagonal are 1 and rest 0

my_identity_matrix=np.eye(4)
print(my_identity_matrix)
```
